```python
 # --- 2. Qué plataforma ha tenido el mayor y menor crecimiento ---

df_plat = df.groupby(['Year', 'Platform'])['Global_Sales'].sum().reset_index()
df_plat = df_plat[df_plat['Year'] > 2010]
df_plat = df_plat[df_plat['Year'] < 2015]

 fig = px.line(df_plat, x = 'Year', y = 'Global_Sales', color = 'Platform')
 fig.show()
```

```python
# Concuciones ejercicio anterior:

# Hasta el año 2015 las plataformas con mayor crecimiento respecto a su mes anterior fueron: PS4 y XOne,
# En cambio las de menor crecimiento fueron: PS3 y X360
```

```python
# --- 3. Qué videojuego ha tenido el mayor y menor crecimiento ---
df_vid = df.groupby(['Year', 'Name'])['Global_Sales'].sum().reset_index()
df_vid = df_vid[df_vid['Year'] > 2010]
df_vid = df_vid[df_vid['Year'] < 2015]

fig = px.line(df_vid, x = 'Year', y = 'Global_Sales', color = 'Name')
fig.show()
```

```python
# Concuciones ejercicio anterior:

# Hasta el año 2015 el videojuego con mayor crecimiento respecto a su mes anterior fue: Grand Theft Auto V,
# En cambio el de menor crecimiento fue: Minecraft
```
